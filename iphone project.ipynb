```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
data = pd.read_csv("apple_products.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
data.describe()
```

# top 10 highest.rated iphone on flipkart in india?

```python
highest_rated = data.sort_values(by = ["Star Rating"],ascending=False)
highest_rated = highest_rated.head(10)
```

```python
print(highest_rated["Product Name"])
```

```python
print(highest_rated)
```

# How many rating do the highest-rated iphones on flipkart have ?


```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated["Number Of Ratings"]
figure = px.bar(highest_rated,x=label,y=counts)
figure.update_layout(width=1000,height=800,xaxis_tickangle=45)
figure.show()
```

# which iphone has the highest number of reviews on flipkart?

```python
data.head()
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated["Number Of Reviews"]
figure = px.bar(highest_rated,x=label,y=counts,title="number of reviews of Highest Rated iphones")
figure.update_layout(width=1000,height=800,xaxis_tickangle=45)
figure.show()
```

# what is the relationship between the sale price of iphones and the number of ratings on Flipkart?

```python
data.head()
```

```python
figure = px.scatter(data_frame = data,x="Number Of Ratings" , y="Sale Price" , size = "Discount Percentage", trendline='ols',title="relationship betwwen sales price and number of rating")
figure.show()
```

# There is a negative linear between the sale price of iphones and the number of ratings. it means iphones with lower sale prices are sold more in india

#what is the relationship between the discount percentage and the ratings of flipkart?

```python
data.head()
```

```python
figure = px.scatter(
    data_frame = data,
    x="Number Of Ratings",
    y = "Discount Percentage",
    size="Sale Price",
    trendline ="ols",
    title= "Relationship between discount percentage and number of ratings of iphones")
figure.show()
```
